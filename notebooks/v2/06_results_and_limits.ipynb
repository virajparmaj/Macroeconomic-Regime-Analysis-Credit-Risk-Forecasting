# 6. Results and limits

**Question this notebook answers**
What can this dataset actually support as a claim, and what can it not? Answer is one table read
from disk, plus a list of limitations each tied to a specific piece of code or a specific
diagnostic.

**What v1 did**
`p2_models.ipynb:3521` claims stacked-LSTM and GRU errors that "outperformed the static
Random-Forest baseline by > 60%". Cell 42 builds the final comparison table from a DataFrame
literal rather than from the run two cells earlier. `:3293` still contains the assistant reasoning
trace that produced the surrounding prose. Each claim is set against the value the notebook itself
printed in `00_audit_of_v1.ipynb` section 8.

**Why that was wrong**
The reported conclusions were written rather than computed. Once a table can contain a typed
constant, no number in the write-up is checkable, and the errors are not small: the audit shows two
of the deep-learning figures out by roughly two orders of magnitude, with the sign of R-squared
reversed.

**What this notebook does instead**
Reads `results/metrics.csv` and renders it. It computes no metrics of its own, and every figure in
the prose is interpolated from that file. Limitations are stated only where a diagnostic in this
repository supports them.

**Predicted effect on results — recorded before running**
The honest headline will be far less impressive than v1's: a model that roughly matches the random
walk and does not reliably beat it, with wide fold-to-fold variation. I expect the Diebold-Mariano
and Clark-West tests to fail to reject equal accuracy against the random walk for the better
models. That is the result, and the write-up should say so plainly rather than searching for a
framing that sounds like success.

**Inputs** → **Outputs**
`results/metrics.csv` → tables and figures in this notebook; figures to `results/figures/`.
