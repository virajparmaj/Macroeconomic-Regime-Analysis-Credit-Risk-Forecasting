# 3. Baselines first

**Question this notebook answers**
What is the number every later model has to beat? Answer is the random walk's out-of-sample MSE on
the one-step-ahead task, established here before any model is fitted.

**What v1 did**
Nothing. The word "baseline" in the naive or random-walk sense appears nowhere in the thirty files
of `notes/`, and no notebook under `notebooks/viraj/` fits one. `p2_models.ipynb` uses a random
forest as its "static baseline" and then compares deep learning against that.

**Why that was wrong**
Without a naive benchmark there is no scale for any reported error. `00_audit_of_v1.ipynb` shows
section 6 scores both on the same data and finds the random forest losing to the random walk — so
the model described as a solid benchmark is beaten by assuming nothing changes. Using a fitted model as the reference point hides that.

**What this notebook does instead**
Fits the random walk, an AR(1) and a train-mean constant first, records them in
`results/metrics.csv` through `src/evaluate.py`, and establishes the bar before any macro variable
is used.

**Predicted effect on results — recorded before running**
The random walk and the AR(1) should be close to each other and clearly ahead of the train-mean
constant. On a Diebold-Mariano test the random walk and AR(1) should not be statistically
distinguishable at this sample size. I expect no later model in notebook 05 to beat the random walk
by a margin that survives that test.

**Inputs** → **Outputs**
`data/v2/panel_pit.parquet` → baseline rows appended to `results/metrics.csv` via
`src.evaluate.write_metrics`.

---
## 3.1 Load features and define the honest task

Load the panel, build features, and state the target explicitly: predict the spread at t+1 using only information available at t.

---
## 3.2 Why persistence is the benchmark

Print the target's autocorrelation at several lags and the implied effective sample size. Look for lag-1 persistence high enough that any benchmark ignoring it is not a serious comparison.

---
## 3.3 Fit the three baselines

Call `src.baselines.all_baselines` on the training block and score on the test block.

---
## 3.4 Baseline metrics table

Report MSE, MAE, R-squared and out-of-sample R-squared against the random walk. Look for the train-mean constant scoring a large negative out-of-sample R-squared.

---
## 3.5 Are the baselines distinguishable from each other?

Run `src.metrics.diebold_mariano` between the random walk and the AR(1). Look at whether the p-value permits calling either one better.

---
## 3.6 Errors over time

Plot baseline forecast errors against the calendar. Look at where the error concentrates, particularly around March and April 2020.

---
## 3.7 Write the baselines to `results/metrics.csv`

Call `src.evaluate.write_metrics`. Every later notebook compares against these rows rather than recomputing them.
